```
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# 1. Initialization
np.random.seed(101)
num_rows = 10000

categories_map = {
    'Electronics': ['Accessories', 'Audio', 'Computers'],
    'Furniture': ['Chairs', 'Tables', 'Desks'],
    'Apparel': ['Shirts', 'Pants', 'Footwear']
}
regions = ['North', 'South', 'East', 'West']
supply_status = ['On-Time', 'Delayed - Transit', 'Delayed - Customs']

data = []
start_date = datetime(2023, 1, 1)

# 2. Data Generation Loop
for i in range(num_rows):
    cat = np.random.choice(list(categories_map.keys()))
    # Deliberately inject ~3% NULLs for SQL cleaning
    sub_cat = np.random.choice(categories_map[cat]) if np.random.rand() > 0.03 else None

    reg = np.random.choice(regions)
    logistics = np.random.choice(supply_status, p=[0.75, 0.15, 0.10])
    base_cost = np.random.uniform(15.0, 400.0)

    if cat == 'Furniture' or logistics != 'On-Time':
        inv_days = int(np.random.normal(loc=115, scale=25))
        markup = np.random.uniform(1.05, 1.35)
    else:
        inv_days = int(np.random.normal(loc=30, scale=12))
        markup = np.random.uniform(1.6, 2.4)

    inv_days = max(1, inv_days)
    disc = np.random.uniform(0.35, 0.65) if inv_days > 85 else np.random.choice([0.0, 0.05])
    unit_price = round(base_cost * markup, 2)
    qty = int(np.random.choice([1, 2, 3, 4, 5]))

    data.append([
        f"ORD-{10000+i}", (start_date + timedelta(days=np.random.randint(0, 365))).strftime('%Y-%m-%d'),
        cat, sub_cat, reg, logistics, qty, round(base_cost, 2), unit_price, round(disc, 2), inv_days
    ])

# 3. Export to CSV
df = pd.DataFrame(data, columns=[
    'order_id', 'order_date', 'category', 'sub_category', 'region', 'supply_chain_status',
    'quantity', 'unit_cost', 'unit_price', 'discount_pct', 'inventory_days'
])
df.to_csv('1_retail_raw_data.csv', index=False)
print("SUCCESS: 1_retail_raw_data.csv generated.")
```
